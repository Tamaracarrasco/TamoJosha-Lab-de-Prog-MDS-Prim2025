````python
## Importación de librerías

# Básicas
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Carga de archivos / sistema
from pathlib import Path
import os
import pickle

# Parquet (engine)
import pyarrow as pa          
import pyarrow.parquet as pq  

# SHAP values
import shap

# Para métricas de desempeño
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix, classification_report
)
````

````python
##### Ruta(s) archivo(s)

### BATCHES 
# BATCH 0
BATCH_0 = "../entrega_3/RESULTADOS_AIRFLOW_MLFLOW/BATCH_0/MLFLOW/ARTIFACTS_BATCH_0/"

# BATCH 1
BATCH_1 = "../entrega_3/RESULTADOS_AIRFLOW_MLFLOW/BATCH_1/MLFLOW/ARTIFACTS_BATCH_1/"

# BATCH 2
BATCH_2  = "../entrega_3/RESULTADOS_AIRFLOW_MLFLOW/BATCH_2/MLFLOW/ARTIFACTS_BATCH_2/"

# BATCH 3
BATCH_3 = "../entrega_3/RESULTADOS_AIRFLOW_MLFLOW/BATCH_3/MLFLOW/ARTIFACTS_BATCH_3/"

# BATCH 4
BATCH_4 = "../entrega_3/RESULTADOS_AIRFLOW_MLFLOW/BATCH_4/MLFLOW/ARTIFACTS_BATCH_4/"

# DATAFRAMES POR BATCHES
DATA_PATH = "../entrega_2/airflow/data/" 
df_last = pd.read_parquet(DATA_PATH + "df_final_latest.parquet")

````

### 📌 Análisis individual [3.0 puntos]

#### 📌 Batch 1 (01/01/25 - 05/01/25) [0.75 puntos]

````python
# Performance batch 1

######## CARGA DEL MODELO #######

model_path = os.path.join(BATCH_1, "model.pkl") 

with open(model_path, "rb") as f:
    model = pickle.load(f)

## EVALUACION 

df = df_last.copy()

# ESTE CONJUNTO ES EL QUE NOS INTERESA

df_test = df[(df["semana"] == "2025-01")].copy()

X_test = df_test.drop(columns='y').copy()
y_test = df_test['y'].copy()

# PREDICCIONES
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1] 

# MÉTRICAS 
print("Accuracy: ", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:   ", recall_score(y_test, y_pred))
print("F1:       ", f1_score(y_test, y_pred))
print("ROC AUC:  ", roc_auc_score(y_test, y_proba))
print(classification_report(y_test, y_pred))
````

````python
#  SHAP values Batch 1

# ============
# Para calcular los SHAP actuales, aplico un submuestreo por temas de costo computacional
# ============
X_batch1 = X_test.copy()

if len(X_batch1) > 20000:
    X_batch1_sample = X_batch1.sample(n=20000, random_state=42)
else:
    X_batch1_sample = X_batch1

print(f"Usando {len(X_batch1_sample)} filas para el análisis SHAP del Batch 1.")

# Convertir columnas categóricas a category (como en el entrenamiento)
for col in ["customer_type", "brand", "category", "sub_category", "segment", "package"]:
    if col in X_batch1_sample.columns:
        X_batch1_sample[col] = X_batch1_sample[col].astype("category")

# ============
# TRANSFORMAR los datos usando solo el ColumnTransformer (sin el clasificador)
# ============
col_transformer = model.named_steps['col_transformer']
X_batch1_transformed = col_transformer.transform(X_batch1_sample)

# Convertir a DataFrame para mantener nombres de features
feature_names = col_transformer.get_feature_names_out()
X_batch1_transformed_df = pd.DataFrame(
    X_batch1_transformed.toarray() if hasattr(X_batch1_transformed, 'toarray') else X_batch1_transformed,
    columns=feature_names
)

# ============
# Calculo los SHAP considerando XGBoost
# (el modelo está dentro de un Pipeline, así que usamos una función de predicción)
# ============
# Extraer solo el clasificador XGBoost del pipeline
xgb_model = model.named_steps['clasificador_xgb']

# Usar TreeExplainer directamente (mucho más rápido)
explainer = shap.TreeExplainer(xgb_model)
shap_values = explainer.shap_values(X_batch1_transformed_df)

# ============
# SHAP summary plot (importancia global de variables en Batch 1)
# ============
plt.figure()
shap.summary_plot(
    shap_values,
    X_batch1_transformed_df,
    show=False
)
plt.title("SHAP summary - Batch 1 (semana 2025-01)")
plt.tight_layout()
plt.show()

# ============
# SHAP dependence plot para 'size'
# ============
# Buscar columna 'size' en las features transformadas
size_col = [col for col in X_batch1_transformed_df.columns if 'size' in col.lower()]
if size_col:
    plt.figure()
    shap.dependence_plot(
        size_col[0], 
        shap_values,
        X_batch1_transformed_df,
        show=False
    )
    plt.title(f"SHAP dependence - {size_col[0]} (Batch 1)")
    plt.tight_layout()
    plt.show()

# ============
# SHAP dependence plot para 'sub_category_GASEOSAS'
# ============
# Buscar columna one-hot encoded de GASEOSAS
gaseosas_col = [col for col in X_batch1_transformed_df.columns if 'gaseosas' in col.lower()]
if gaseosas_col:
    plt.figure()
    shap.dependence_plot(
        gaseosas_col[0], 
        shap_values,
        X_batch1_transformed_df,
        show=False
    )
    plt.title(f"SHAP dependence - {gaseosas_col[0]} (Batch 1)")
    plt.tight_layout()
    plt.show()
````

**Análisis Comparativo semanas 1 a 4**

En el análisis comparativo de performance a través de las cuatro semanas de enero de dos mil veinticinco se observa un patrón consistente que ayuda a entender cómo se comporta el sistema de monitoreo, drift y reentrenamiento en producción. En todos los batches se mantiene el mismo volumen de observaciones (ciento setenta y tres mil quinientas treinta y cinco filas por semana) y, según el comportamiento del pipeline, en las cuatro corridas el check_drift_task detectó cambios suficientes para gatillar reentrenamiento, de modo que cada semana se evalúa un modelo actualizado, con hiperparámetros y pesos ligeramente distintos pero siempre bajo la misma lógica de arquitectura y umbral de decisión cercano a cero coma ocho. Esto hace comparables las métricas entre semanas, pero obliga a interpretar las variaciones como resultado de una combinación entre drift en los datos entrantes y la respuesta adaptativa del modelo reentrenado.

Al examinar las métricas de forma conjunta, emerge una separación clara entre dos grupos. Por un lado, recall y ROC-AUC se mantienen en un rango alto y con baja volatilidad a lo largo de todo el período. El recall oscila en una franja muy estrecha, entre cero coma setecientos sesenta aproximadamente en el Batch 3 y cero coma setecientos setenta y dos en el Batch 4, con una mejora neta de alrededor de uno coma tres por ciento entre la primera y la última semana. De manera similar, el ROC-AUC se mantiene siempre por sobre cero coma ochocientos cincuenta, partiendo en torno a cero coma ochocientos cincuenta y cuatro en el Batch 1, alcanzando su máximo cercano a cero coma ochocientos sesenta y nueve en el Batch 3 y cerrando el mes en torno a cero coma ochocientos sesenta en el Batch 4, lo que representa una ganancia neta respecto del punto de partida. Esta combinación de alta sensibilidad y alta capacidad discriminativa, prácticamente estable en el tiempo, indica que el modelo conserva de manera robusta su habilidad para ordenar correctamente a los clientes por probabilidad de compra y para capturar a la gran mayoría de los compradores reales, pese a los cambios semanales en la distribución de los datos y a los reentrenamientos sucesivos.

En contraste, las métricas de clasificación binaria que dependen más directamente del umbral de decisión (accuracy, precision y F1) exhiben un patrón de “curva en campana” mucho más marcado. Las tres mejoran de forma simultánea entre el Batch 1 y el Batch 2 y luego se deterioran en los Batch 3 y 4 hasta nivelar o incluso empeorar levemente respecto del valor inicial. La accuracy pasa de aproximadamente cero coma ochocientos cuatro en el Batch 1 a un máximo cercano a cero coma ochocientos veinticinco en el Batch 2, desciende ligeramente a cero coma ochocientos dieciocho en el Batch 3 y culmina en torno a cero coma setecientos noventa y nueve en el Batch 4, quedando algo por debajo del punto de partida. La precision sigue el mismo patrón con mayor volatilidad relativa pues sube desde valores cercanos a cero coma cero ochenta y tres en el Batch 1 hasta alrededor de cero coma ciento once en el Batch 2 (un aumento relativo cercano al treinta por ciento), para luego descender a cerca de cero coma cero noventa y ocho en el Batch 3 y finalizar en torno a cero coma cero ochenta en el Batch 4, que es incluso ligeramente inferior al valor inicial. Dado que el recall permanece casi invariante, el F1 refleja casi exactamente esta dinámica de la precision pues se incrementa desde alrededor de cero coma ciento cincuenta en el Batch 1 a aproximadamente cero coma ciento noventa y tres en el Batch 2, cae a cerca de cero coma ciento setenta y cuatro en el Batch 3 y termina en torno a cero coma ciento cuarenta y seis en el Batch 4. Es decir, el mejor equilibrio entre captura y selectividad se observa en la segunda semana, mientras que la cuarta semana muestra el peor compromiso, con F1 y precision en sus niveles más bajos del período.

Esta divergencia entre las métricas de ranking (ROC-AUC) y sensibilidad (recall), por un lado, y las métricas de clasificación dura (precision, F1 y en menor medida accuracy), por otro, sugiere que el mecanismo de reentrenamiento está logrando preservar la “calidad estructural” del modelo (su capacidad para asignar probabilidades bien calibradas y para separar positivos y negativos en el espacio de puntajes), pero que la conversión de esas probabilidades a decisiones binarias mediante un umbral fijo deja de ser óptima a medida que la distribución de las probabilidades se desplaza con el tiempo. Que el ROC-AUC mejore netamente entre el Batch 1 y el Batch 3, mientras precision y F1 se deterioran en los Batch 3 y 4, es consistente con un escenario en el que el modelo ordena cada vez mejor las instancias, pero una proporción creciente de observaciones queda por encima del umbral o en su vecindad, generando más falsos positivos sin que ello implique una pérdida de capacidad discriminativa subyacente. En otras palabras, el drift que el pipeline está capturando y corrigiendo vía reentrenamiento parece ser principalmente distributivo (cambios en frecuencias, mezcla de segmentos y combinaciones de categorías) más que conceptual en el sentido de invertir la lógica de las relaciones entre variables y respuesta.

La lectura por batch refuerza esta interpretación. El Batch 1 representa el punto de partida, con un modelo ya reentrenado que exhibe alta sensibilidad y AUC elevado, pero con precision baja y un F1 modesto, reflejando desde el inicio el compromiso entre no perder compradores y tolerar un número importante de falsos positivos en un contexto de clase minoritaria extrema. El Batch 2 constituye claramente la mejor semana del período pues todas las métricas, salvo el recall que se mantiene prácticamente igual, mejoran de modo visible; accuracy, precision y F1 alcanzan sus máximos y el AUC experimenta un salto positivo, lo que indica que ese reentrenamiento en particular logró un modelo más nítido tanto en ranking como en clasificación dura bajo el mismo umbral. En el Batch 3 se observa una situación intermedia pues el ROC-AUC alcanza su mejor valor de todo el mes y el recall se mantiene alto, pero accuracy, precision y F1 comienzan a retroceder, anticipando que el umbral fijo deja de ser tan eficiente para mantener la selectividad pese a que el modelo sigue separando bien las clases. Finalmente, el Batch 4 representa la semana más desafiante pues el AUC sigue siendo alto y el recall incluso alcanza su máximo, pero accuracy, precision y F1 caen a sus mínimos del período, con valores muy similares o algo peores que los observados en la primera semana. Operativamente esto implica que el sistema sigue encontrando a la mayoría de los compradores, pero a costa del nivel más alto de sobrealertas de todo el mes.

El análisis comparativo muestra que el pipeline con detección de drift y reentrenamiento semanal logra mantener una capacidad discriminativa robusta y una sensibilidad elevada de manera sostenida a lo largo de enero, lo que es coherente con el objetivo de no perder oportunidades de compra. Sin embargo, también pone en evidencia que las métricas de selectividad y balance (precision y F1) son mucho más inestables y que las mejoras puntuales obtenidas tras ciertos reentrenamientos (particularmente en el Batch 2) no se consolidan, sino que se erosionan en las semanas sucesivas. Esto sugiere que, además de monitorizar drift y actualizar el modelo, sería recomendable complementar el esquema actual con una revisión periódica del umbral de decisión o con una estrategia adaptativa de umbrales, y con monitoreos específicos de calibración de probabilidades, de modo que la excelente capacidad de ranking que el modelo demuestra de forma estable se traduzca de manera más consistente en decisiones binarias con mejor equilibrio entre captura y sobrealertas a lo largo del tiempo.
