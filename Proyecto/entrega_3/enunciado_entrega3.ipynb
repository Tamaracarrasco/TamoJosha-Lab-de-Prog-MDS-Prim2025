````python
## Importación de librerías

# Básicas
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Carga de archivos / sistema
from pathlib import Path
import os
import pickle

# Parquet (engine)
import pyarrow as pa          
import pyarrow.parquet as pq  

# SHAP values
import shap

# Para métricas de desempeño
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix, classification_report
)
````

````python
##### Ruta(s) archivo(s)

### BATCHES 
# BATCH 0
BATCH_0 = "../entrega_3/RESULTADOS_AIRFLOW_MLFLOW/BATCH_0/MLFLOW/ARTIFACTS_BATCH_0/"

# BATCH 1
BATCH_1 = "../entrega_3/RESULTADOS_AIRFLOW_MLFLOW/BATCH_1/MLFLOW/ARTIFACTS_BATCH_1/"

# BATCH 2
BATCH_2  = "../entrega_3/RESULTADOS_AIRFLOW_MLFLOW/BATCH_2/MLFLOW/ARTIFACTS_BATCH_2/"

# BATCH 3
BATCH_3 = "../entrega_3/RESULTADOS_AIRFLOW_MLFLOW/BATCH_3/MLFLOW/ARTIFACTS_BATCH_3/"

# BATCH 4
BATCH_4 = "../entrega_3/RESULTADOS_AIRFLOW_MLFLOW/BATCH_4/MLFLOW/ARTIFACTS_BATCH_4/"

# DATAFRAMES POR BATCHES
DATA_PATH = "../entrega_2/airflow/data/" 
df_last = pd.read_parquet(DATA_PATH + "df_final_latest.parquet")

````

### 📌 Análisis individual [3.0 puntos]

#### 📌 Batch 1 (01/01/25 - 05/01/25) [0.75 puntos]

````python
# Performance batch 1

######## CARGA DEL MODELO #######

model_path = os.path.join(BATCH_1, "model.pkl") 

with open(model_path, "rb") as f:
    model = pickle.load(f)

## EVALUACION 

df = df_last.copy()

# ESTE CONJUNTO ES EL QUE NOS INTERESA

df_test = df[(df["semana"] == "2025-01")].copy()

X_test = df_test.drop(columns='y').copy()
y_test = df_test['y'].copy()

# PREDICCIONES
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1] 

# MÉTRICAS 
print("Accuracy: ", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:   ", recall_score(y_test, y_pred))
print("F1:       ", f1_score(y_test, y_pred))
print("ROC AUC:  ", roc_auc_score(y_test, y_proba))
print(classification_report(y_test, y_pred))
````

````python
#  SHAP values Batch 1

# Para calcular los SHAP actuales, aplico un submuestreo por temas de costo computacional

X_batch1 = X_test.copy()

if len(X_batch1) > 20000:
    X_batch1_sample = X_batch1.sample(n=20000, random_state=42)
else:
    X_batch1_sample = X_batch1

print(f"Usando {len(X_batch1_sample)} filas para el análisis SHAP del Batch 1.")

# Convertir columnas categóricas a category (como en el entrenamiento)
for col in ["customer_type", "brand", "category", "sub_category", "segment", "package"]:
    if col in X_batch1_sample.columns:
        X_batch1_sample[col] = X_batch1_sample[col].astype("category")


# TRANSFORMAR los datos usando solo el ColumnTransformer (sin el clasificador)

col_transformer = model.named_steps['col_transformer']
X_batch1_transformed = col_transformer.transform(X_batch1_sample)

# Se convierte a dataframe para mantener los nombres de features
feature_names = col_transformer.get_feature_names_out()
X_batch1_transformed_df = pd.DataFrame(
    X_batch1_transformed.toarray() if hasattr(X_batch1_transformed, 'toarray') else X_batch1_transformed,
    columns=feature_names
)

# Calculo los SHAP considerando XGBoost

# (el modelo está dentro de un Pipeline, así que usamos una función de predicción)

# Extraer solo el clasificador XGBoost del pipeline
xgb_model = model.named_steps['clasificador_xgb']

explainer = shap.TreeExplainer(xgb_model)
shap_values = explainer.shap_values(X_batch1_transformed_df)

# SHAP summary plot (importancia global de variables en Batch 1)
plt.figure()
shap.summary_plot(
    shap_values,
    X_batch1_transformed_df,
    show=False
)

plt.title("SHAP summary - Batch 1 (semana 2025-01)")
plt.tight_layout()
plt.show()

# SHAP dependence plot para 'size'
# Buscamos columna 'size' en las features transformadas
size_col = [col for col in X_batch1_transformed_df.columns if 'size' in col.lower()]
if size_col:
    plt.figure()
    shap.dependence_plot(
        size_col[0], 
        shap_values,
        X_batch1_transformed_df,
        show=False
    )
    plt.title(f"SHAP dependence - {size_col[0]} (Batch 1)")
    plt.tight_layout()
    plt.show()


# SHAP dependence plot para 'sub_category_GASEOSAS'
# Buscamos columna one-hot encoded de GASEOSAS
gaseosas_col = [col for col in X_batch1_transformed_df.columns if 'gaseosas' in col.lower()]
if gaseosas_col:
    plt.figure()
    shap.dependence_plot(
        gaseosas_col[0], 
        shap_values,
        X_batch1_transformed_df,
        show=False
    )
    plt.title(f"SHAP dependence - {gaseosas_col[0]} (Batch 1)")
    plt.tight_layout()
    plt.show()
````

````python
# Analisis comparativo de la performace de los batches semana a semana


# COMPARATIVO DE PERFORMANCE A TRAVÉS DE LAS SEMANAS (BATCHES 1-4)

# Lista para almacenar las métricas de cada batch
results = []

df = df_last.copy()

# BATCH 1: Semana 2025-01
print("="*60)
print("EVALUANDO BATCH 1 - Semana 2025-01")
print("="*60)

model_path = os.path.join(BATCH_1, "model.pkl")
with open(model_path, "rb") as f:
    model_batch1 = pickle.load(f)

df_test_1 = df[df["semana"] == "2025-01"].copy()
X_test_1 = df_test_1.drop(columns='y').copy()
y_test_1 = df_test_1['y'].copy()

y_pred_1 = model_batch1.predict(X_test_1)
y_proba_1 = model_batch1.predict_proba(X_test_1)[:, 1]

results.append({
    'Batch': 'Batch 1',
    'Semana': '2025-01',
    'Período': '01-01 a 05-01',
    'N_muestras': len(y_test_1),
    'Accuracy': accuracy_score(y_test_1, y_pred_1),
    'Precision': precision_score(y_test_1, y_pred_1),
    'Recall': recall_score(y_test_1, y_pred_1),
    'F1': f1_score(y_test_1, y_pred_1),
    'ROC_AUC': roc_auc_score(y_test_1, y_proba_1)
})

print(f"F1: {results[-1]['F1']:.4f} | ROC-AUC: {results[-1]['ROC_AUC']:.4f}")

# BATCH 2: Semana 2025-02

print("\n" + "="*60)
print("EVALUANDO BATCH 2 - Semana 2025-02")
print("="*60)

model_path = os.path.join(BATCH_2, "model.pkl")
with open(model_path, "rb") as f:
    model_batch2 = pickle.load(f)

df_test_2 = df[df["semana"] == "2025-02"].copy()
X_test_2 = df_test_2.drop(columns='y').copy()
y_test_2 = df_test_2['y'].copy()

y_pred_2 = model_batch2.predict(X_test_2)
y_proba_2 = model_batch2.predict_proba(X_test_2)[:, 1]

results.append({
    'Batch': 'Batch 2',
    'Semana': '2025-02',
    'Período': '06-01 a 11-01',
    'N_muestras': len(y_test_2),
    'Accuracy': accuracy_score(y_test_2, y_pred_2),
    'Precision': precision_score(y_test_2, y_pred_2),
    'Recall': recall_score(y_test_2, y_pred_2),
    'F1': f1_score(y_test_2, y_pred_2),
    'ROC_AUC': roc_auc_score(y_test_2, y_proba_2)
})

print(f"F1: {results[-1]['F1']:.4f} | ROC-AUC: {results[-1]['ROC_AUC']:.4f}")

# BATCH 3: Semana 2025-03

print("\n" + "="*60)
print("EVALUANDO BATCH 3 - Semana 2025-03")
print("="*60)

model_path = os.path.join(BATCH_3, "model.pkl")
with open(model_path, "rb") as f:
    model_batch3 = pickle.load(f)

df_test_3 = df[df["semana"] == "2025-03"].copy()
X_test_3 = df_test_3.drop(columns='y').copy()
y_test_3 = df_test_3['y'].copy()

y_pred_3 = model_batch3.predict(X_test_3)
y_proba_3 = model_batch3.predict_proba(X_test_3)[:, 1]

results.append({
    'Batch': 'Batch 3',
    'Semana': '2025-03',
    'Período': '13-01 a 19-01',
    'N_muestras': len(y_test_3),
    'Accuracy': accuracy_score(y_test_3, y_pred_3),
    'Precision': precision_score(y_test_3, y_pred_3),
    'Recall': recall_score(y_test_3, y_pred_3),
    'F1': f1_score(y_test_3, y_pred_3),
    'ROC_AUC': roc_auc_score(y_test_3, y_proba_3)
})

print(f"F1: {results[-1]['F1']:.4f} | ROC-AUC: {results[-1]['ROC_AUC']:.4f}")

# BATCH 4: Semana 2025-04
print("\n" + "="*60)
print("EVALUANDO BATCH 4 - Semana 2025-04")
print("="*60)

model_path = os.path.join(BATCH_4, "model.pkl")
with open(model_path, "rb") as f:
    model_batch4 = pickle.load(f)

df_test_4 = df[df["semana"] == "2025-04"].copy()
X_test_4 = df_test_4.drop(columns='y').copy()
y_test_4 = df_test_4['y'].copy()

y_pred_4 = model_batch4.predict(X_test_4)
y_proba_4 = model_batch4.predict_proba(X_test_4)[:, 1]

results.append({
    'Batch': 'Batch 4',
    'Semana': '2025-04',
    'Período': '20-01 a 26-01',
    'N_muestras': len(y_test_4),
    'Accuracy': accuracy_score(y_test_4, y_pred_4),
    'Precision': precision_score(y_test_4, y_pred_4),
    'Recall': recall_score(y_test_4, y_pred_4),
    'F1': f1_score(y_test_4, y_pred_4),
    'ROC_AUC': roc_auc_score(y_test_4, y_proba_4)
})

print(f"F1: {results[-1]['F1']:.4f} | ROC-AUC: {results[-1]['ROC_AUC']:.4f}")

# TABLA COMPARATIVA

print("\n" + "="*80)
print("TABLA COMPARATIVA DE MÉTRICAS POR BATCH")
print("="*80)

df_results = pd.DataFrame(results)
print(df_results.to_string(index=False))

# Guardar tabla
df_results.to_csv("metricas_comparativas_batches.csv", index=False)
print("\n Tabla guardada como 'metricas_comparativas_batches.csv'")

# GRÁFICOS DE TENDENCIA

# Configuración de estilo
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 10)

# Crear subplots
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Evolución del Performance del Modelo a través de las Semanas (Enero 2025)', 
             fontsize=16, fontweight='bold')

metrics = ['Accuracy', 'Precision', 'Recall', 'F1', 'ROC_AUC']
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

# Gráfico 1: Todas las métricas juntas
ax = axes[0, 0]
for i, metric in enumerate(metrics):
    ax.plot(df_results['Batch'], df_results[metric], 
            marker='o', linewidth=2, markersize=8, 
            label=metric, color=colors[i])
ax.set_xlabel('Batch', fontsize=11)
ax.set_ylabel('Score', fontsize=11)
ax.set_title('Todas las Métricas', fontsize=12, fontweight='bold')
ax.legend(loc='best')
ax.grid(True, alpha=0.3)
ax.set_ylim([0, 1])

# Gráficos individuales para cada métrica
positions = [(0, 1), (0, 2), (1, 0), (1, 1), (1, 2)]

for idx, (metric, color, pos) in enumerate(zip(metrics, colors, positions)):
    ax = axes[pos[0], pos[1]]
    ax.plot(df_results['Batch'], df_results[metric], 
            marker='o', linewidth=2.5, markersize=10, 
            color=color, label=metric)
    
    # Agregar valores en cada punto
    for i, (batch, value) in enumerate(zip(df_results['Batch'], df_results[metric])):
        ax.text(i, value + 0.02, f'{value:.3f}', 
               ha='center', va='bottom', fontsize=9)
    
    ax.set_xlabel('Batch', fontsize=11)
    ax.set_ylabel('Score', fontsize=11)
    ax.set_title(f'{metric}', fontsize=12, fontweight='bold')
    ax.grid(True, alpha=0.3)
    ax.set_ylim([max(0, df_results[metric].min() - 0.1), 
                 min(1, df_results[metric].max() + 0.1)])

plt.tight_layout()
plt.savefig('tendencia_metricas_batches.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n Gráfico guardado como 'tendencia_metricas_batches.png'")


# GRÁFICO DE BARRAS COMPARATIVO

fig, ax = plt.subplots(figsize=(14, 7))

x = range(len(df_results))
width = 0.15

for i, (metric, color) in enumerate(zip(metrics, colors)):
    offset = width * (i - 2)
    ax.bar([p + offset for p in x], df_results[metric], 
           width, label=metric, color=color, alpha=0.8)

ax.set_xlabel('Batch', fontsize=12, fontweight='bold')
ax.set_ylabel('Score', fontsize=12, fontweight='bold')
ax.set_title('Comparación de Métricas por Batch', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(df_results['Batch'])
ax.legend(loc='best')
ax.grid(True, alpha=0.3, axis='y')
ax.set_ylim([0, 1.1])

plt.tight_layout()
plt.savefig('comparacion_barras_batches.png', dpi=300, bbox_inches='tight')
plt.show()

print("Gráfico guardado como 'comparacion_barras_batches.png'")


# ANÁLISIS DE TENDENCIAS

print("\n" + "="*80)
print("ANÁLISIS DE TENDENCIAS")
print("="*80)

for metric in metrics:
    valores = df_results[metric].values
    tendencia = "MEJORA" if valores[-1] > valores[0] else "DETERIORO" if valores[-1] < valores[0] else "ESTABLE"
    cambio_absoluto = valores[-1] - valores[0]
    cambio_porcentual = (cambio_absoluto / valores[0]) * 100 if valores[0] != 0 else 0
    
    print(f"\n{metric}:")
    print(f" - Valor inicial (Batch 1): {valores[0]:.4f}")
    print(f" - Valor final (Batch 4):   {valores[-1]:.4f}")
    print(f" - Tendencia: {tendencia}")
    print(f" - Cambio absoluto: {cambio_absoluto:+.4f}")
    print(f" - Cambio porcentual: {cambio_porcentual:+.2f}%")
    
    # desviación estándar
    std = df_results[metric].std()
    print(f" Desviacion estandar: {std:.4f}")

# TABLA RESUMEN DE CAMBIOS

print("\n" + "="*80)
print("RESUMEN DE CAMBIOS (Batch 1 → Batch 4)")
print("="*80)

cambios = []
for metric in metrics:
    valores = df_results[metric].values
    cambios.append({
        'Métrica': metric,
        'Batch_1': valores[0],
        'Batch_4': valores[-1],
        'delta_Absoluto': valores[-1] - valores[0],
        'delta_Porcentual': ((valores[-1] - valores[0]) / valores[0] * 100) if valores[0] != 0 else 0,
        'std': df_results[metric].std()
    })

df_cambios = pd.DataFrame(cambios)
print(df_cambios.to_string(index=False))

df_cambios.to_csv("resumen_cambios_batches.csv", index=False)
print("\n Tabla guardada como 'resumen_cambios_batches.csv'")
````

**Análisis Comparativo semanas 1 a 4**

En el análisis comparativo de performance a lo largo de las cuatro semanas de enero de 2025 se observa un patrón consistente que permite comprender cómo se comporta el sistema de monitoreo, drift y reentrenamiento en producción. En todos los batches se mantiene el mismo volumen de observaciones (173_535 filas por semana) y, según el comportamiento del pipeline, en las cuatro corridas el `check_drift_task` detectó cambios suficientes para gatillar reentrenamiento. De este modo, cada semana se evalúa un modelo actualizado, con hiperparámetros y pesos ligeramente distintos pero siempre bajo la misma arquitectura y un umbral de decisión cercano a 0.8. Esto hace comparables las métricas entre semanas, aunque obliga a interpretar las variaciones como resultado de la combinación entre el drift en los datos entrantes y la respuesta adaptativa del modelo reentrenado.

Al examinar las métricas en conjunto, emerge una separación clara entre dos grupos. Por un lado, recall y ROC-AUC se mantienen en un rango alto y con baja desviación estándar a lo largo del período. El recall oscila en una franja muy estrecha, entre aproximadamente 0.760 en el Batch 3 y 0.772 en el Batch 4, con una mejora neta cercana al 1.3% entre la primera y la última semana. De manera similar, el ROC-AUC se mantiene siempre por sobre 0,850: parte en torno a 0.854 en el Batch 1, alcanza su máximo cercano a 0.869 en el Batch 3 y cierra el mes alrededor de 0.860 en el Batch 4, lo que representa una ganancia respecto del punto de partida. Esta combinación de alta sensibilidad y alta capacidad discriminativa, prácticamente estable en el tiempo, indica que el modelo conserva de manera robusta su habilidad para ordenar correctamente a los clientes por probabilidad de compra y para capturar a la gran mayoría de los compradores reales, pese a los cambios semanales en la distribución de los datos y a los reentrenamientos sucesivos.

En contraste, las métricas de clasificación binaria más dependientes del umbral de decisión (accuracy, precision y F1) exhiben un patrón mucho más marcado de “curva en campana”. Las tres mejoran simultáneamente entre el Batch 1 y el Batch 2, pero luego se deterioran en los Batch 3 y 4 hasta nivelarse o incluso quedar levemente por debajo del valor inicial. La accuracy pasa de aproximadamente 0.804 en el Batch 1 a un máximo cercano a 0.825 en el Batch 2, desciende a 0.818 en el Batch 3 y culmina en torno a 0.799 en el Batch 4, quedando algo por debajo del punto de partida. La precision sigue el mismo patrón, aunque con mayor desviación relativa: sube desde valores cercanos a 0.083 en el Batch 1 hasta alrededor de 0.111 en el Batch 2 (un aumento relativo cercano al 30%), para luego descender a cerca de 0.098 en el Batch 3 y cerrar en torno a 0.080 en el Batch 4, que es incluso ligeramente inferior al valor inicial. Como el recall permanece casi invariante, el F1 refleja casi exactamente la dinámica de la precision: se incrementa desde alrededor de 0.150 en el Batch 1 hasta aproximadamente 0.193 en el Batch 2, cae a cerca de 0.174 en el Batch 3 y termina en torno a 0.146 en el Batch 4. Es decir, el mejor equilibrio entre captura y selectividad se observa en la segunda semana, mientras que la cuarta semana muestra el peor, con F1 y precision en sus niveles más bajos del período.

Esta divergencia entre las métricas de ranking (ROC-AUC) y sensibilidad (recall), por un lado, y las métricas de clasificación dura (precision, F1 y, en menor medida, accuracy), por otro, sugiere que el mecanismo de reentrenamiento está logrando preservar la “calidad estructural” del modelo: su capacidad para asignar probabilidades bien calibradas y separar positivos y negativos en el espacio de puntajes. Sin embargo, la conversión de esas probabilidades en decisiones binarias mediante un umbral fijo deja de ser óptima a medida que la distribución de probabilidades se desplaza con el tiempo. Que el ROC-AUC mejore netamente entre el Batch 1 y el Batch 3, mientras precision y F1 se deterioran en los Batch 3 y 4, es consistente con un escenario en el que el modelo ordena cada vez mejor las instancias, pero una proporción creciente de observaciones queda por encima del umbral o muy cerca de él, generando más falsos positivos sin que ello implique una pérdida de capacidad discriminativa. En otras palabras, el drift que el pipeline está capturando y corrigiendo vía reentrenamiento parece ser principalmente distributivo (cambios en frecuencias, mezcla de segmentos y combinaciones de categorías), más que conceptual en el sentido de invertir relaciones fundamentales entre variables y respuesta.

La lectura por batch refuerza esta interpretación. El Batch 1 representa el punto de partida, con un modelo ya reentrenado que exhibe alta sensibilidad y AUC elevado, pero con precision baja y un F1 modesto, reflejando desde el inicio el compromiso entre no perder compradores y tolerar un número importante de falsos positivos en un contexto de clase extremadamente minoritaria. El Batch 2 constituye la mejor semana del período: todas las métricas, salvo el recall que se mantiene casi igual, mejoran de forma visible; accuracy, precision y F1 alcanzan sus máximos y el AUC experimenta un salto positivo, lo que indica que ese reentrenamiento produjo un modelo más nítido tanto en ranking como en clasificación dura bajo el mismo umbral. En el Batch 3 se observa una situación intermedia: el ROC-AUC alcanza su mejor valor del mes y el recall se mantiene alto, pero accuracy, precision y F1 comienzan a retroceder, anticipando que el umbral fijo deja de ser eficiente para mantener la selectividad, pese a que el modelo sigue separando bien las clases. Finalmente, el Batch 4 representa la semana más desafiante: el AUC sigue siendo alto y el recall incluso alcanza su máximo, pero accuracy, precision y F1 caen a sus mínimos del período, con valores similares o algo peores que los observados en la primera semana. Operativamente, esto implica que el sistema continúa detectando a la mayoría de los compradores, pero a costa del nivel más alto de sobrealertas de todo el mes.

En síntesis, el análisis comparativo muestra que el pipeline con detección de drift y reentrenamiento semanal mantiene una capacidad discriminativa robusta y una sensibilidad elevada de manera sostenida durante enero, lo que es coherente con el objetivo de no perder oportunidades de compra. Sin embargo, también evidencia que las métricas de selectividad y balance (precision y F1) son mucho más inestables y que las mejoras puntuales obtenidas tras ciertos reentrenamientos (en particular en el Batch 2) no se consolidan, sino que se empeoran en semanas posteriores. Esto sugiere que, además de monitorizar drift y actualizar el modelo, sería recomendable complementar el esquema con una revisión periódica del umbral de decisión o con una estrategia de umbrales adaptativos, junto con monitoreos específicos de calibración de probabilidades, de modo que la excelente capacidad de ranking que el modelo demuestra de forma estable se traduzca de manera más consistente en decisiones binarias con mejor equilibrio entre captura y sobrealertas a lo largo del tiempo.
