En el competitivo universo de las bebidas gaseosas, la empresa **SodAI Drinks 🥤** ha logrado destacarse por su creatividad, diversidad de productos y enfoque centrado en el cliente. Ofrece una extensa gama de bebidas carbonatadas que abarca distintos segmentos del mercado: desde productos premium en presentaciones sofisticadas, hasta gaseosas accesibles para el consumo masivo, disponibles en diversos tamaños y tipos de envases. 

La compañía opera en múltiples regiones y zonas, sirviendo a una variedad de puntos de venta que incluyen desde tiendas de conveniencia y minimarkets hasta el canal frío tradicional. Cada tipo de cliente tiene sus particularidades: algunos reciben entregas hasta 4 veces por semana, mientras que otros son visitados por la fuerza de ventas solo una vez semanalmente. Esta diversidad de perfiles representa tanto una oportunidad como un desafío comercial: ¿cómo saber qué productos tienen más chances de ser comprados por cada cliente en un momento dado?

Con el objetivo de aumentar la facturación de forma inteligente y mejorar la eficiencia de su estrategia de ventas, **SodAI Drinks** decide crear una nueva célula interna de innovación: el equipo **Deep Drinkers 🤖**, cuyo propósito es aplicar ciencia de datos para anticiparse a las necesidades del cliente y potenciar el negocio desde una perspectiva basada en información.

El corazón de esta iniciativa es el desarrollo de un sistema predictivo personalizado para cada cliente. Para ello, **Deep Drinkers** convoca a un equipo de Data Scientists y especialistas en *machine learning* con una misión clara: construir un modelo predictivo que, cada semana, pueda estimar la probabilidad de compra de cada producto del portafolio para cada cliente activo.

El modelo deberá tener en cuenta múltiples factores, incluyendo:
- **Tipo de cliente**, ej. "TIENDA DE CONVENIENCIA", "MINIMARKET".
- **Frecuencia de entregas y visitas**, indicadores del nivel de actividad comercial.
- **Ubicación geográfica** (por región y zona).
- **Preferencias históricas de consumo**, inferidas por patrones de compra anteriores.
- **Características del producto**, como marca, categoría, segmento, tipo de envase y tamaño

El objetivo final es que, **cada semana**, se genere una tabla de productos priorizados: para cada cliente, un listado de productos ordenado por su probabilidad estimada de compra. Esta información será enviada al equipo comercial, que podrá usarla en call center, para incrementar las chances de concretar ventas al ofrecer justo lo que el cliente probablemente quiere comprar.

Este proyecto representa un cambio de paradigma en la forma en que **SodAI Drinks** gestiona su fuerza de ventas: de un enfoque reactivo y generalista, a uno proactivo, basado en datos y profundamente personalizado. Así, la empresa no solo espera aumentar su rentabilidad, sino también construir relaciones más sólidas con sus clientes, ofreciéndoles recomendaciones más relevantes y oportunas.

Para lograr lo anterior, el equipo **Deep Drinkers** contará con los siguientes conjuntos de datos, junto a sus respectivos atributos:

- **Datos transaccionales** (`transacciones.parquet`): contiene el historial de compras realizadas por los clientes.
	- `customer_id`: identificador único del cliente que realizó la compra.
	- `product_id`: identificador único del producto comprado.
	- `purchase_date`: fecha en que se realizó la transacción.
	- `order_id`: identificar de la orden de su pedido.
	- `items`: cantidad de bultos comprados por cliente en aquella transacción.

- **Datos de clientes** (`clientes.parquet`): incluye las características de cada cliente.
	- `customer_id`: identificador único del cliente.
	- `region_id`: identificador de la región geográfica donde se encuentra el cliente.
	- `customer_type`: tipo de cliente según el canal comercial, por ejemplo, “TIENDA DE CONVENIENCIA”.
	- `Y`: coordenada geográfica de latitud.
	- `X`: coordenada geográfica de longitud.
	- `num_deliver_per_week`: cantidad de entregas semanales que recibe el cliente.
	- `num_visit_per_week`: frecuencia de visitas de la fuerza de ventas por semana.

- **Datos de productos** (`productos.parquet`): describe las características de los productos del portafolio.
	- `product_id`: identificador único del producto.
	- `brand`: marca comercial del producto.
	- `category`: categoría general del producto, como “BEBIDAS CARBONATADAS”.
	- `sub_category`: subcategoría dentro de la categoría principal, por ejemplo, “GASEOSAS”.
	- `segment`: segmento de mercado al que pertenece el producto, como “PREMIUM”.
	- `package`: tipo de envase del producto.
	- `size`: tamaño del producto en litros.

*Abstract*

- *Problema.*
El objetivo del proyecto es diseñar y desplegar un sistema de recomendación semanal que, para cada cliente activo y cada producto del portafolio, prediga la probabilidad de compra en la semana siguiente (t+1) y, a partir de ese puntaje, clasifique si ocurrirá o no la compra. La predicción se basa en el comportamiento histórico de cada cliente (patrones de recompra y estacionalidad), en sus atributos operativos (tipo de cliente, frecuencia de visitas y entregas, ubicación) y en las características de los productos (marca, categoría, segmento, envase y tamaño). Con estas probabilidades generamos una lista priorizada por cliente (Top-N SKU) para activar al equipo comercial, por ejemplo, en call center, de modo que puedan ofrecer lo más probable de comprar en ese momento, aumentando la conversión y la eficiencia del contacto. El sistema se ejecuta cada semana, cubre múltiples regiones y tipos de canal, y produce dos salidas complementarias con un puntaje continuo (propensión) para ordenar y focalizar la oferta y una etiqueta binaria para seguimiento y evaluación operativa. Al final, lo que buscamos es pasar de una venta reactiva y generalista a una gestión proactiva y personalizada, guiada por datos, que anticipe necesidades y eleve ingresos con la misma capacidad comercial.

- *Datos de entrada.*
Disponemos de tres fuentes integradas a nivel transaccional: (i) transacciones, con customer_id, product_id, order_id, purchase_date (fecha sin hora) e items (cantidades que pueden ser fraccionarias por conversión de bultos y eventualmente negativas por devoluciones); (ii) clientes, con atributos operacionales y geoespaciales como customer_type (canal), region_id, zone_id, coordenadas X,Y, y frecuencias num_deliver_per_week y num_visit_per_week; y (iii) productos, con metadatos comerciales (brand, category, sub_category, segment, package, size). La clave natural en transacciones es el trío (customer_id, product_id, order_id) y la granularidad es línea de pedido; un mismo order_id puede contener múltiples productos. La cobertura temporal efectiva va de 2024-01-01 a 2024-12-31 y la resolución temporal útil es diaria, aunque para el modelamiento se proyecta a panel semanal. En la auditoría detectamos duplicados totales (filas idénticas) y duplicados parciales (mismo cliente–producto–orden/fecha con items distintos, típicos de ajustes/devoluciones); para asegurar consistencia consolidamos por suma neta por clave y filtramos casos con neto ≤ 0 (anulación/devolución completa). También observamos esparsidad marcada del cubo cliente×producto×semana (muchos pares sin compra la mayoría de las semanas), alto desbalance entre clases (no-compra ≫ compra) y alta cardinalidad en variables categóricas (marcas, segmentos, tipos de cliente), lo que condiciona el preprocesamiento (codificación robusta y control de categorías raras) y la selección de métricas/modelos. Entonces, contamos con datos transaccionales reales, con atributos ricos de cliente y producto, un año completo de historia y particularidades operativas (devoluciones, ajustes, frecuencias de visita/entrega) que requieren normalización temporal y control de calidad para usarlos en predicción semanal de compra.

- *Métrica de evaluación.*
Usaremos métricas que reflejen el carácter altamente desbalanceado del problema observado en el EDA (las compras semanales por par cliente–producto son una minoría*del total; en validación, la clase positiva ronda el 2–3%). Por eso, la métrica principal será el AUCPR (Average Precision / área bajo la curva Precision–Recall), que mide la calidad del ranking cuando hay muchísimos negativos y evita la falsa sensación de desempeño que puede dar el AUROC en escenarios desbalanceados. Para decidir el umbral operativo de clasificación (cuando haga falta una etiqueta 0/1), seleccionaremos el que maximiza el F1 de la clase positiva sobre validación, privilegiando capturar compras reales (recall) sin perder totalmente la precisión. Reportaremos también Accuracy y AUROC solo como referencia, advirtiendo que aquí pueden ser engañosas (por ejemplo, predecir siempre “no compra” ya supera el 97% de exactitud). Finalmente, porque el entregable es una lista priorizada por cliente, complementaremos con métricas de negocio Top-N como Precision@N / Hit-Rate@N y Lift, que miden cuánto mejor es ofrecer los N SKUs con mayor probabilidad frente a ofrecer al azar; estas métricas conectan directamente con el uso del call center y permiten sintonizar N según la capacidad operativa.

- *Modelos y transformaciones.*
El proyecto se implementó con un pipeline reproducible en scikit-learn que primero sanea y estructura los datos y luego entrena los modelos. Partimos corrigiendo duplicados totales y parciales (cuando un mismo cliente–producto–orden–fecha aparece con distintas cantidades, consolidamos por suma y descartamos netos ≤ 0 por corresponder a anulaciones o devoluciones completas). A continuación, transformamos el histórico transaccional en un panel semanal anclado al lunes donde agregamos por cliente–producto–semana, reindexamos semanas faltantes para distinguir ausencia de compra de ausencia de registro, definimos compra_actual (compra en la semana t) y construimos la variable objetivo ("y") como compra en la semana t+1. Para evitar leakage, los splits se realizan por la semana del target (entrenamiento enero–septiembre, validación octubre–noviembre, prueba diciembre), de modo que las filas con etiqueta t+1 nunca usen información de esa misma semana como feature.

Sobre ese panel aplicamos un ColumnTransformer con imputación por mediana en numéricas (coordenadas X, Y, size, entregas y visitas por semana, purchased_count, compra_actual) y por moda en categóricas; escalamos numéricas con StandardScaler para favorecer la convergencia de modelos lineales y codificamos categóricas con One-Hot Encoding usando min_frequency y handle_unknown="ignore" para controlar la cardinalidad y la esparsidad. Excluimos del modelo columnas de identificación y temporales que inducen fuga (customer_id, product_id, week_t, week_t_plus_1, etiquetas de semana), conservando únicamente señales operativas de cliente (tipo, región, zona, visitas y entregas) y de producto (marca, categoría/subcategoría, segmento, envase, tamaño), además de las señales de comportamiento semanal ya agregadas.

Evaluamos como baselines Regresión Logística y SGDClassifier por su rapidez, estabilidad y buena calibración en espacios esparsos, además de un Decision Tree para contrastar no linealidades simples. El modelo final es XGBoost, seleccionado por capturar interacciones no lineales sin feature engineering pesado, escalar bien con OHE disperso y manejar el desbalance mediante scale_pos_weight (relación negativos/positivos); lo entrenamos con tree_method="hist" y early stopping sobre validación. Los hiperparámetros se ajustaron con Optuna y, ya con el mejor conjunto, reentrenamos en train+val y fijamos el umbral de decisión con la validación para maximizar F1 de la clase positiva. El modelo entrega tanto la probabilidad de compra en t+1 (usada para priorizar y rankear Top-N SKUs por cliente-semana) como la clasificación binaria requerida para seguimiento operativo.

-*Resultados y conclusiones.*

Esta combinación con alto recall y con precision baja, es consistente con un espacio de millones de candidatos y pocas compras semanales pues priorizamos no “perder” compras potenciales y utilizamos la probabilidad como puntaje de ranking para que el call center accione sobre un Top-N por cliente. En ese régimen de uso (no con un umbral global), la precision práctica aumenta sustantivamente al enfocarse solo en los SKUs mejor posicionados de cada cliente, lo que habilita listas priorizadas que son más efectivas que reglas generales o contacto aleatorio.

Las conclusiones principales son que (i) el modelo produce un ranking útil y accionable que satisface el requerimiento de probabilidades semanales, (ii) XGBoost es la mejor alternativa entre los modelos evaluados bajo métricas robustas al desbalance y con corte temporal, y (iii) la operación por Top-N, más que un umbral fijo, es la forma correcta de explotar el output del modelo.
```python
def generate_january_2025(df,
                          date_col="purchase_date",
                          order_id_col="order_id",
                          items_col="items",
                          noise_factor=0.50,   # ruido
                          seed=42):

    np.random.seed(seed)
    df = df.copy()

    # Asegurar datetime
    df[date_col] = pd.to_datetime(df[date_col])

    # Tomamos weekday
    df["weekday"] = df[date_col].dt.weekday

    # Semana objetivo 2025
    target_start = pd.Timestamp("2025-01-01")
    target_end = pd.Timestamp("2025-01-07")
    target_week = list(pd.date_range(target_start, target_end, freq="D"))

    # Mapeo weekday -> weekday
    def map_to_2025(row):
        wd = row["weekday"]
        for d in target_week:
            if d.weekday() == wd:
                return d
        return target_start  # fallback

    df[date_col] = df.apply(map_to_2025, axis=1)

    # Regenerar order_id único
    df[order_id_col] = np.arange(100000, 100000 + len(df))

    # Se agrega ruido a items
    # Ruido = N(0, noise_factor * items)
    noise = np.random.normal(loc=0,
                             scale=np.abs(df[items_col]) * noise_factor)

    df[items_col] = df[items_col] + noise

    # Opcional: redondear a 2 decimales
    df[items_col] = df[items_col].round(2)

    # Flags para tracking
    df["is_synthetic"] = True
    df["synthetic_reason"] = "shift_jan2024_to_jan2025"

    # Dropear auxiliares
    df = df.drop(columns=["weekday"])

    return df
```

```python
from pandas.api.types import is_numeric_dtype

def show_count_and_sample(df, mask, cols=None, sample_n=10, title=""):
    cnt = int(mask.sum())
    print(f"\n{title} -> {cnt:,} fila(s)")
    if cnt > 0:
        display(df.loc[mask, cols].head(sample_n) if cols else df.loc[mask].head(sample_n))

## En esta celda, tratamos de verificar la validez de la información,
## por ejemplo: notamos que habían clientes que se repetian el número de orden
## fecha, producto pero no "items"

# 0) Duplicados por clave transaccional 
cols_clave = ["customer_id", "product_id", "order_id", "purchase_date"]
mask_dups_clave = transacciones.duplicated(subset=cols_clave, keep=False)

show_count_and_sample(transacciones, mask_dups_clave, cols=cols_clave + ["items"], 
                      title="(0) Duplicados por clave (customer_id, product_id, order_id, purchase_date)")

# 1) items imposibles: ≤ 0 o excesivos (configurable)

ITEMS_MAX = 1_000  
mask_items_no_pos = transacciones["items"] <= 0
mask_items_muy_altos = transacciones["items"] > ITEMS_MAX

show_count_and_sample(transacciones, mask_items_no_pos, 
                      cols=["customer_id","product_id","order_id","purchase_date","items"], 
                      title="(1) items ≤ 0 (devoluciones totales/anulaciones o error)")

show_count_and_sample(transacciones, mask_items_muy_altos, 
                      cols=["customer_id","product_id","order_id","purchase_date","items"], 
                      title=f"(1b) items > {ITEMS_MAX} (posible carga errónea)") # Son clientes mayoristas?

# 2) Integridad de llaves entre tablas

mask_cli_fk = ~transacciones["customer_id"].isin(clientes["customer_id"])
mask_prod_fk = ~transacciones["product_id"].isin(productos["product_id"])

show_count_and_sample(transacciones, mask_cli_fk, 
                      cols=["customer_id","order_id","purchase_date"], 
                      title="(2) customer_id de transacciones que NO existen en clientes")

show_count_and_sample(transacciones, mask_prod_fk, 
                      cols=["product_id","order_id","purchase_date"], 
                      title="(2b) product_id de transacciones que NO existen en productos")

# 3) Un order_id asociado a múltiples clientes 
cli_por_order = transacciones.groupby("order_id")["customer_id"].nunique()
mask_order_multi_cli = transacciones["order_id"].isin(cli_por_order[cli_por_order > 1].index)

show_count_and_sample(transacciones, mask_order_multi_cli, 
                      cols=["order_id","customer_id","product_id","purchase_date","items"], 
                      title="(3) order_id con múltiples customer_id (inconsistencia grave)")

# 4) Fechas inválidas o que estén fuera de rango
fecha_min, fecha_max = transacciones["purchase_date"].min(), transacciones["purchase_date"].max()
mask_fecha_na = transacciones["purchase_date"].isna()

show_count_and_sample(transacciones, mask_fecha_na, title="(4) purchase_date nulo(s)")
print(f"Rango de fechas: {fecha_min} → {fecha_max}")

# 5) Coordenadas fuera de rango 
if {"Y","X"}.issubset(clientes.columns):
    mask_lat_inv = (clientes["Y"] < -90) | (clientes["Y"] > 90)
    mask_lon_inv = (clientes["X"] < -180) | (clientes["X"] > 180)
    show_count_and_sample(clientes, mask_lat_inv, cols=["customer_id","Y","X"], 
                          title="(5) Latitudes fuera de rango (-90,90)")
    show_count_and_sample(clientes, mask_lon_inv, cols=["customer_id","Y","X"], 
                          title="(5b) Longitudes fuera de rango (-180,180)")

# 6) Frecuencias semanales fuera de rango (Clientes)

for col, name in [("num_deliver_per_week","entregas/semana"),
                  ("num_visit_per_week","visitas/semana")]:

    if col in clientes.columns and is_numeric_dtype(clientes[col]):
        mask_neg   = clientes[col] < 0
        mask_gt7   = clientes[col] > 7
        mask_float = (clientes[col] % 1 != 0)  # si esperas enteros
        show_count_and_sample(clientes, mask_neg, cols=["customer_id", col], 
                              title=f"(6) {col} < 0 (inválido)")
        show_count_and_sample(clientes, mask_gt7, cols=["customer_id", col], 
                              title=f"(6b) {col} > 7 (inválido si es semanal)")
        show_count_and_sample(clientes, mask_float, cols=["customer_id", col], 
                              title=f"(6c) {col} no entero (revisar si deben ser enteros)")

# 7) Tamaño de producto inválido 

if "size" in productos.columns and is_numeric_dtype(productos["size"]):
    SIZE_MAX = 20 
    mask_size_le0 = productos["size"] <= 0
    mask_size_gt  = productos["size"] > SIZE_MAX

    show_count_and_sample(productos, mask_size_le0, cols=["product_id","brand","package","size"], 
                          title="(7) size ≤ 0 (inválido)")
    show_count_and_sample(productos, mask_size_gt, cols=["product_id","brand","package","size"], 
                          title=f"(7b) size > {SIZE_MAX}L (posible error)")

# 8) Categorías faltantes o vacías en productos

for c in ["brand","category","sub_category","segment","package"]:
    if c in productos.columns:
        mask_na = productos[c].isna() | (productos[c].astype(str).str.strip() == "")
        show_count_and_sample(productos, mask_na, cols=["product_id", c], 
                              title=f"(8) {c} nulo o vacío")

# 9) Items con demasiada granularidad decimal 
# Ejemplo: si los packs vienen en tercios (0.3333) o sextos (0.1667), esto puede ser OK.
# Aquí sólo marcamos valores con >4 decimales significativos por si sospechas problemas de conversión.

mask_many_decimals = (transacciones["items"].round(4) != transacciones["items"])
show_count_and_sample(transacciones, mask_many_decimals, 
                      cols=["customer_id","product_id","order_id","purchase_date","items"], 
                      title="(9) items con >4 decimales (revisar si es esperado)")
```

```python
# Selección de variables útiles para modelado

# Eliminación de posibles valores negativos residuales
df = df[df["items"] > 0]

# Convertir categóricas a string si aún no lo están
for c in ["customer_type", "brand", "category", "zone_id"]:
    df[c] = df[c].astype(str)

print("Dataset final listo para modelado:", df.shape)
```

```python
# Creacion variable objetivo y variables explicativas

df_ = df.copy()
df_["purchase_date"] = pd.to_datetime(df_["purchase_date"], errors="coerce")
df_ = df_.dropna(subset=["purchase_date"])

# Resample semanal por par (cliente, producto)
panel = (
    df_.set_index("purchase_date")
       .groupby(["customer_id","product_id"])
       .resample("W-MON")                    # semanas ancladas a lunes
       .size()
       .rename("purchased_count")
       .reset_index()
)

# Variables solicitadas
panel.rename(columns={"purchase_date":"week_t"}, inplace=True)
panel["compra_o_no"] = (panel["purchased_count"] > 0).astype(int)
panel = panel.sort_values(["customer_id","product_id","week_t"])

# Target: compra en la próxima semana (y en t+1)
panel["y"] = panel.groupby(["customer_id","product_id"])["compra_o_no"].shift(-1)
panel = panel.dropna(subset=["y"]).copy()
panel["y"] = panel["y"].astype(int)

# Semana t+1
panel["week_t_plus_1"] = panel["week_t"] + pd.Timedelta(days=7)

# Etiqueta "semana" estilo YYYY-ww (ISO)
iso = panel["week_t"].dt.isocalendar()
panel["semana"] = iso.year.astype(str) + "-" + iso.week.astype(str).str.zfill(2)

# Etiqueta para t+1
iso1 = panel["week_t_plus_1"].dt.isocalendar()
panel["semana_siguiente_str"] = iso1.year.astype(str) + "-" + iso1.week.astype(str).str.zfill(2)

# Dataset final
df_copy = panel[[
    "customer_id","product_id",
    "week_t","week_t_plus_1","semana","semana_siguiente_str",
    "purchased_count","compra_o_no","y"
]].reset_index(drop=True)
```

```python
# Construcción de la base:

customers = (
    df["customer_id"]
    .drop_duplicates()
    .sort_values()
    .to_numpy()
)
products = (
    df["product_id"]
    .drop_duplicates()
    .sort_values()
    .to_numpy()
)

# Semanas 
weeks = (
    pd.to_datetime(df_copy["week_t"])
    .drop_duplicates()
    .sort_values()
    .to_list()
)



frames = []
n_prod = len(products)
n_cust = len(customers)

for wk in weeks:
    wk_ts = pd.Timestamp(wk)
    wk_plus1 = wk_ts + pd.Timedelta(days=7)

    # Etiquetas YYYY-ww para semana actual 
    iso0 = wk_ts.isocalendar()
    iso1 = wk_plus1.isocalendar()
    semana_str = f"{iso0.year}-{int(iso0.week):02d}"
    semana_next_str = f"{iso1.year}-{int(iso1.week):02d}"

    # Producto cartesiano clientes x productos SOLO para esta semana
    base = pd.DataFrame({
        "customer_id": np.repeat(customers, n_prod),
        "product_id": np.tile(products, n_cust),
        "week_t": wk_ts,
        "week_t_plus_1": wk_plus1,
        "semana": semana_str,
        "semana_siguiente_str": semana_next_str
    })

    # Filas reales de esa semana desde df_copy 
    wk_rows = df_copy.loc[
        df_copy["week_t"].eq(wk_ts),
        ["customer_id","product_id","purchased_count","compra_o_no","y"]
    ]

    # Merge por semana (left = todos los pares; right = observados)
    merged = base.merge(wk_rows, on=["customer_id","product_id"], how="left")

    # Relleno con 0 donde no hubo compra 
    for c in ["purchased_count","compra_o_no","y"]:
        merged[c] = merged[c].fillna(0).astype(int)

    frames.append(merged)

# Se concatenan todas las semanas

df_full = pd.concat(frames, ignore_index=True)

# Acá añadimos las features de clientes y productos

cols_cli = ["customer_id","customer_type","X","Y","zone_id",
            "region_id","num_deliver_per_week","num_visit_per_week"]
cols_cli = [c for c in cols_cli if c in df.columns]  # por si falta alguna

cols_prod = ["product_id","brand","category","sub_category",
             "segment","package","size"]
cols_prod = [c for c in cols_prod if c in df.columns]

# Dimensión clientes (1 fila por customer_id)
dim_clientes = (
    df[cols_cli]
      .drop_duplicates(subset=["customer_id"])
      .reset_index(drop=True)
)

# Dimensión productos (1 fila por product_id)
dim_productos = (
    df[cols_prod]
      .drop_duplicates(subset=["product_id"])
      .reset_index(drop=True)
)

# Merge de dimensiones (LEFT, para no perder combinaciones)
df_final = (
    df_full
    .merge(dim_clientes, on="customer_id", how="left")
    .merge(dim_productos, on="product_id", how="left")
)

# Check de dimensiones 
print("Filas df_copy (panel observado):", len(df_copy))
print("Filas df_full (cartesiano por semana):", len(df_full))
print("Filas df_final (con features):", len(df_final))
print("Positivos (y=1):", int(df_final["y"].sum()))
print("Tasa y=1:", df_final["y"].mean())

```

```python
# Desarrolle aquí su código

from sklearn.impute import SimpleImputer 
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer


drop_cols = ["customer_id", "product_id", "week_t", "week_t_plus_1",
             "semana", "semana_siguiente_str", "semana_num"]

num_cols = [
    "X","Y","size",
    "num_deliver_per_week","num_visit_per_week",
    "purchased_count","compra_o_no"]

cat_cols = ["customer_type","segment","brand","category","sub_category","zone_id","region_id"]

# Pipelines
num_pipeline = Pipeline([
    ("imp", SimpleImputer(strategy="median")),   # mediana para robustez
    ("sc", StandardScaler())
])

cat_pipeline = Pipeline([
    ("imp", SimpleImputer(strategy="most_frequent")),
    ("ohe", OneHotEncoder(handle_unknown="ignore", sparse_output=True))
])

col_transformer = ColumnTransformer(
    transformers=[
        ("drop_ids", "drop", drop_cols),
        ("num",  num_pipeline, num_cols),
        ("cat",  cat_pipeline, cat_cols),
    ],
    remainder="drop",
    verbose_feature_names_out=False
)
```

*Justificaciones*

El preprocesamiento se implementa como un pipeline reproducible que garantiza el mismo flujo de limpieza y transformación cada vez que se entrena o se infiere. Partimos desde el panel semanal ya depurado (deduplicados totales/parciales consolidados por suma y netos ≤ 0 removidos) y construido con información hasta la semana t; allí definimos purchased_count y compra_o_no (compra en t), y la etiqueta y como compra en t+1. Para evitar fuga temporal, retiramos del set de entrenamiento identificadores y columnas de fecha/semana (customer_id, product_id, week_t, week_t_plus_1, etiquetas de semana), de modo que las features de una fila en t no usen señales de t+1.

Dentro del pipeline, primero aplicamos imputación de valores nulos diferenciada por tipo-> en numéricas (X, Y, size, num_deliver_per_week, num_visit_per_week, purchased_count, compra_o_no) usamos mediana, que es robusta a asimetrías y outliers; en categóricas (customer_type, segment, brand, category, sub_category, zone_id, region_id) usamos moda para mantener consistencia semántica. A continuación, estandarizamos numéricas con StandardScaler para mejorar la convergencia de modelos lineales y mantener comparabilidad de magnitudes; para modelos de árboles esto no es estrictamente necesario, pero permite probar distintas familias de modelos sin cambiar el resto del flujo. Las categóricas se convierten a numéricas mediante One-Hot Encoding con handle_unknown="ignore" (evita fallos ante categorías nuevas en inferencia) y salida dispersa para controlar memoria; si la cardinalidad crece, el mismo pipeline admite fijar min_frequency para agrupar categorías raras sin reescribir código. No incluimos discretización de continuas por defecto porque XGBoost y árboles capturan no linealidades sin necesidad de binning; no obstante, el pipeline es modular y permite insertar, como variante, discretizaciones supervisadas (p. ej. KBinsDiscretizer) si aportan valor en validación. Respecto de valores atípicos, la combinación de mediana + modelos de árboles ya ofrece robustez; si el análisis de residuos lo requiriese, el pipeline permite incorporar winsorization (recorte por cuantiles) o RobustScaler como paso intercambiable.

La orquestación de pasos se realiza con `ColumnTransformer`y en paralelo “drop” de columnas que causan fuga/identificadores, imputación+escalado para numéricas y imputación+OHE para categóricas; el orden queda fijado (imputar → escalar/encodear) y todo se ajusta únicamente con datos de entrenamiento, preservando validación y prueba para estimar generalización.

*Explicacion modelso elegidos*
1) Regresión logística (baseline, lineal): modelo lineal que estima la probabilidad con una función sigmoide. Caso 3 (media): ABARROTES + GASEOSAS + tamaño mediano + 1–2 entregas/semana → probabilidad intermedia. Razón: GASEOSAS sube, pero size no tan chico y logística moderada frenan; pequeñas mejoras (más entregas o formato más chico) subirían la probabilidad.

2) *Atributo más importante y por qué tiene sentido*

El más importante por gain es sub_category_GASEOSAS. Tiene sentido de negocio pues es la subcategoría de mayor rotación, con ciclos cortos de reposición y alta respuesta a promociones; por eso los árboles la usan como primer split para explicar gran parte de la variación de compra en t+1.

3) *Interacciones relevantes*

sub_category_GASEOSAS × size: dentro de gaseosas, formatos pequeños se recompran antes que los grandes (ciclo más corto).

customer_type_ABARROTES × num_deliver_per_week: en retail tradicional, más entregas -> más oportunidades de reposición -> mayor probabilidad.

Geografía (X,Y) × logística: la misma frecuencia de entrega no impacta igual en todas las zonas (cobertura/costos/lead-time); los árboles ajustan por ubicación.

4) Riesgos de sesgo y mitigación

Hacia categorías: favorece GASEOSAS (más gain y más positivos en el histórico). Mitigar: cuotas/Top-N por familia, métricas por categoría, o modelos por clúster de categorías.

Hacia tipos de cliente: señal fuerte en ABARROTES. Mitigar: monitorear AUCPR/F1 por segmento y umbral por segmento.

Hacia zonas: X,Y con datos desbalanceados pueden penalizar áreas. Mitigar: revisar métricas por región, enriquecer con variables de cobertura y ajustar umbrales por zona.

A modo de reflexión final entonces:

El objetivo fue predecir, semana a semana, la probabilidad de que cada cliente compre cada SKU y, en paralelo, clasificar si habrá compra en la semana siguiente (t+1). Partimos de transacciones, clientes y productos; limpiamos duplicados totales y “parciales” (mismo cliente–producto–orden/fecha con cantidades distintas) consolidando items por suma y descartando netos ≤ 0; estandarizamos fechas y construimos un panel semanal anclado al lunes que incluye explícitamente semanas sin compra (clase negativa), lo que evita sesgos por ausencia de registros. Definimos la variable objetivo como compra en t+1 por par (cliente, producto) y separamos temporalmente 2024 en entrenamiento, validación y prueba (70/20/10) sin leakage. El pipeline de preprocesamiento usó imputación (mediana/moda), One-Hot Encoding con manejo de categorías raras y estandarización para numéricas; con ColumnTransformer aplicamos cada transformación a su bloque correspondiente, garantizando reproducibilidad y consistencia entre splits.

En métricas, el problema es altamente desbalanceado, por lo que evaluamos con F1 de la clase positiva y AUCPR (como referencia de ranking), reportando además precisión, recall y exactitud. En validación, los modelos lineales y de árbol sirvieron de baseline Regresión Logística (class_weight=balanced) entregó F1+(≈0.18), SGDClassifier (~0.19) y DecisionTree (~0.20). El mejor desempeño lo obtuvo XGBoost (tree_method=hist, scale_pos_weight), con F1+(≈0.218) y buena exactitud (~0.865). Posteriormente aplicamos Optuna para ajustar hiperparámetros del modelo y del preprocesamiento (p. ej., min_frequency del OHE). Ese trial incrementó fuertemente el recall positivo (≈0.806) pero redujo la precisión (≈0.073), con F1+(≈0.134) y exactitud más baja (~0.749): el cambio más determinante fue un min_frequency demasiado alto que colapsó muchas categorías infrecuentes y diluyó señal útil; además, la optimización usó la F1 de predict() con umbral 0.5, distinto del umbral operativo que luego seleccionamos en validación. En suma, el mejor clasificador fue el XGBoost “base” (sin esa agresiva compactación de categorías), complementado con selección de umbral en validación para alinear el punto de operación a negocio; con él generamos la tabla de probabilidades por cliente–semana y el ranking Top-N listo para el call center.

Las decisiones de data cleaning y del panel semanal completo fueron críticas al introducir explícitamente los ceros, el modelo aprendió verdaderos negativos por cliente–SKU–semana, mejorando la calibración de probabilidades y la discriminación. El scale_pos_weight en XGBoost ayudó a subir recall sin explotar el costo computacional. En cambio, la agresividad en min_frequency dentro del OHE tuvo un impacto negativo porque removió información específica de marcas/segmentos raros pero predictivos. También observamos que la estandarización beneficia a modelos lineales (Regresión/SGD) y es neutra para árboles/boosting; y que tunar el umbral (en vez de fijar 0.5) es tan importante como ajustar hiperparámetros cuando el objetivo es priorizar contactos. En Optuna, los hiperparámetros del modelo que más movieron la aguja fueron learning_rate y n_estimators (con max_depth y min_child_weight modulando el sesgo/varianza), pero el mayor giro vino del preprocesamiento (OHE), lo que refuerza que modelo y features deben optimizarse en conjunto.

Las principales lecciones son tres. Primero, en problemas semanales y desbalanceados, la elección de métrica y del umbral debe estar guiada por el uso operativo optimizar F1+ y seleccionar umbral en validación produce rankings accionables más útiles que optimizar exactitud. Segundo, el preprocesamiento importa tanto como el modelo consolidar duplicados, construir negativos explícitos y regular la cardinalidad categórica con cuidado (en vez de colapsarla en exceso) eleva el signal-to-noise. Tercero, la validación debe ser temporal y, si es posible, cruzada en el tiempo (TimeSeriesSplit) para robustecer la selección. Con más tiempo/recursos, mejoraríamos (i) features de secuencia (recencia, frecuencia, ventana móvil 4–8–12 semanas, streaks, estacionalidad por semana del año y dummies de calendario), (ii) encoders supervisados con CV (target/James-Stein) para categorías de alta cardinalidad, (iii) calibración de probabilidades (isotónica/Platt) y curvas coste-beneficio para fijar umbrales por capacidad del call center, (iv) modelos adicionales como LightGBM/CatBoost y validación con early stopping y mayor presupuesto de trials, y (v) una evaluación de Top-N por cliente y AUCPR específica por tipo de cliente/segmento para afinar la priorización. Entonces, el sistema cumple el objetivo de entrega probabilidades por cliente–SKU–semana y un ranking Top-N que el equipo comercial puede usar de inmediato; y deja trazado un camino claro de mejoras iterativas que, sin cambiar la arquitectura, deberían traducirse directamente en mayor conversión y facturación.
